```python
# --- Probability with Sets Example ---

#Set-up
#Totals sample space will be 100
sample_space = set(range(1, 101))

#Countries using >50% renewable energy will be 60 
renewable = set(range(1,61)) 

#Countries using <10% fossil fuel will be 40
unrenewable = set(range(1,41))

#Intersection, countries that meet both criteria
both = renewable & unrenewable #& is the intersection code symbol

# -----------
#Basic Probabilities Set
P_A = len(renewable) / len(sample_space) #Finds the number of countries using renewable within sample space threshold
P_B = len(unrenewable) / len(sample_space)
P_A_and_B = len(both) / len(sample_space) #Finds the number of countries meeting both criteria 

#Conditional Probability; Finding the probability of event A happening (A | B)
P_A_Given_B = P_A_and_B / P_B 
            #Note to self, dividing it is rescaling the sample space. Giving a proper probability result, 
            #subtracting will usually give out a result that is negative (nonsensical), greater than 1 (impossible for probability)

# ----------
#Printing Results
print(f"The number of countries using renewable energy are: {P_A}")
print(f"The number of countries using fossil fuels are: {P_B}")
print(f"The number of countries that meet both the criteria are {both}")

print("\n ---- \n")
print(f"Given the number of countries using renewable energy, the probability of it occurring within the sample set of 100 countries are: {P_A_Given_B}")
```
