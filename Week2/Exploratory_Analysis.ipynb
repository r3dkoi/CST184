```python
!pip3 install scipy
#Exploratory Data Analysis: Histograms and KDE

import pandas as pd
import matplotlib.pyplot as plt

#Load cleaned data
df = pd.read_csv("High_energy_consumption_appliances.csv")

#Making a Histogram
df["Energy Consumption (kWh)"].plot(kind="hist", bins=10, edgecolor="black")
plt.title("Histogram of Energy Consumption (kWh)")
plt.xlabel("Energy Consumption")
plt.grid(True)
plt.show()
```
