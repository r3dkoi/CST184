```python
!pip3 install scipy
#Exploratory Data Analysis: Histograms and KDE

import pandas as pd
import matplotlib.pyplot as plt

#Load cleaned data
df = pd.read_csv("smart_home_energy_consumption_large.csv")

#Making a Histogram
df["Energy Consumption (kWh)"].plot(kind="hist", bins=10, edgecolor="black")
plt.title("Histogram of Energy Consumption (kWh)")
plt.xlabel("Energy Consumption")
plt.grid(True)
plt.show()

#KDE plot
df["Energy Consumption (kWh)"].plot(kind="density", linewidth=2)
plt.title("Density plot of Energy Consumption")
plt.xlabel("Energy Consumption")
plt.grid(True)
plt.show()
```
