```python
# --- Bayes' Theorem Example ---

#Set up

#Probability of power blackout (P(A))
P_blackout =  0.10

#Probability of heatwave on blackout days (P(B) | A)
P_heatwave_given_blackout = 0.80

#Overall probability of heatwave (P(B))
P_heatwave = 0.30

# Bayes Theorem, finding the probability of days having a blackout during a heatwave [P(A|B) = (P (B | A) x P (A)] / P (B)
P_blackout_given_heatwave = (P_heatwave_given_blackout * P_blackout) / P_heatwave

#Displaying result
print("The probability of a blackout occurring during a heatwave is:", round(P_blackout_given_heatwave, 3))
```
