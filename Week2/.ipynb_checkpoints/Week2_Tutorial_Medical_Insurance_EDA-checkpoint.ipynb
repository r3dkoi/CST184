```python
#Loading and Inspecting the Data
import pandas as pd
df = pd.read_csv("insurance.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
#STEP 2
#Filtering Data Based on Probability (age)
#This is used to explore more meaningful subsets of the data, e.g
#the age and smoker columns can lead to some insight on medical insurance usage

#What's the number in a specific age range?
low_age_range = df[df["age"] < 20]
middle_age_range = df[df["age"] > 50]
high_age_range = df[df["age"] > 60]
print("\n --- \n")
print("Below are the age ranges with medical insurance charges")
print(f"The number of people with medical insurance charges that are equal to or less than the age 20 are: {len(low_age_range)}.")
print(f"The number of people with medical insurance charges that are equal or greater than 50 but below 60 are: {len(middle_age_range)}.")
print(f"The number of people with medical insurance charges that are equal or greater than the age of 60 but below 70 are: {len(high_age_range)}.")

print("\n --- \n")
print("Below are the numbers of smokers vs non-smokers in this dataset")
#Number of smokers and non-smokers in this dataset
yes_smoker = df[df["smoker"] == "yes"]
no_smoker = df[df["smoker"] == "no"]
print(f"The number of smokers in this dataset are: {len(yes_smoker)}.")
print(f"The number of non-smokers in this dataset are: {len(no_smoker)}.")
```
