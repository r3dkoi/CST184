```python
import pandas as pd
df = pd.DataFrame({
    'date': ['2024-01-01', '2024-01-01', '2024-01-02', '2024-01-02'],
    'city': ['NYC', 'LA', 'NYC', 'LA'],
    'temperature': [32, 65, 35, 68]
})
# Pivot
pivoted = df.pivot(index='date', columns='city', values='temperature')
pivoted
```

```python
import pandas as pd

# Sample DataFrame with MultiIndex columns
df = pd.DataFrame({
    ('A', 'one'): [1, 2],
    ('A', 'two'): [3, 4],
    ('B', 'one'): [5, 6],
    ('B', 'two'): [7, 8]
})

# Stack the innermost column level
stacked = df.stack()
stacked
```
