```r
#Step 1 Preparing Data

#Defining parameters
co2 <- runif(50, min = 2, max= 30)
temperature_anomaly <- 0.6 * co2 + 
    rnorm(50, mean = 0, sd = 3)

#Creating a dataframe
climate_df <- data.frame(
    co2,
    temperature_anomaly
    )

head(climate_df)
```

```r
#Step 2 fitting the model using lm()
model <- lm(temperature_anomaly ~ co2, data = climate_df)
```

```r
#View the model summary
summary(model)
```

```r
#Key outputs to look out for
slope <- 0.54290
intercept <- 0.56049
p_value <- 0.000000000000006531
adjusted_r <- 0.7154

#std errors
co2_error <- 0.04873
intercept_error <- 0.90971
```
