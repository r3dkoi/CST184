```r
#Goal: Learn how to apply a two-sample t-test and calculate a 95% confidence interval in R using a simplified real-world 
#scenario aligned with SDG 3 (Health) and SDG 13 (Climate Action).
```

```r
#Scenario
#You work for a health research agency. You want to test whether people living in high air pollution areas have higher asthma rates than those in low pollution areas.
#You collect data from two samples (30 individuals each) and run a two-sample t-test in R.
```

```r
#Step 1 - Create Mock Data
#Using R to simulate asthma case rates per 1000 people in two regions
# Simulate asthma data (synthetic for learning)
set.seed(42)

low_pollution <- rnorm(30, mean = 11, sd = 3)   # Group A
high_pollution <- rnorm(30, mean = 18, sd = 4)  # Group B

# Combine into a dataframe
asthma_df <- data.frame(
  rate = c(low_pollution, high_pollution),
  group = c(rep("Low Pollution", 30), rep("High Pollution", 30))
)

head(asthma_df)
```

```r
#Step 2 - Run a t-test
t_result <- t.test(rate ~ group, data = asthma_df)
print(t_result)
```

```r
print("From the results above, the p-value is lower than 0.05. This indicates the data is statistically significant. There is a difference in asthma rates between groups.")
```

```r
print("95% confident that the true difference in asthma rates between high and low pollution groups fall between 4.24 and 8.37 cases.")
```

```r
#Step 3 Visualising the Groups
library(ggplot2)

ggplot(asthma_df, aes(x = group, y = rate, fill = group)) +
  geom_boxplot() +
  labs(
    title = "Asthma Rates by Pollution Exposure",
    y = "Asthma Rate (per 1000 people)",
    x = "Region"
  ) +
  theme_minimal()
```
