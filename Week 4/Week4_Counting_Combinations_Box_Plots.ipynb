```python
"""You are working for a global sustainability agency developing a dashboard to monitor CO₂ emissions from renewable energy
strategies across continents. You’ve received a dataset containing emissions per region and energy type. Your task is to calculate how
many strategy options exist, create new features, and visualize how emissions vary by region-energy pairing."""
```

```python
import pandas as pd
df = pd.read_csv("week4_energy_emissions.csv")
df.head()
```

```python
import math
from itertools import product
regions = ['Asia', 'Africa', 'Europe', 'South America', 'North America', 'Australia'] #This doesn't alter the CSV data btw
types = df["Energy_Type"].unique()
# How many unique combinations?
print("Combinations (nCr):", math.comb(len(regions) + len(types), 2))
# How many region-energy pairs?
pairs = list(product(regions, types))
print("Region × Energy combinations:", len(pairs))
```

```python
df["Region_Energy"] = df["Region"] + "_" + df["Energy_Type"]
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x="Region_Energy", y="CO2_Emissions")
plt.title("CO₂ Emissions by Region + Energy Type")
plt.xticks(rotation=45)
plt.ylabel("CO₂ Emissions (tonnes)")
plt.tight_layout()
plt.show()
```
