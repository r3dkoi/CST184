```python
"""You’re part of a sustainability analytics team evaluating renewable energy policies across regions. Your team has collected data on
CO₂ emissions for different region + energy type combinations. Your task is to use box plots to compare the distribution of emissions
across these groupings. This will help inform which policy combinations are most effective"""
```

```python
# Week 4 – Box Plot Visualisation for Categorical Features
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import random
# Step 1: Simulated dataset
regions = ['Asia', 'Africa', 'Europe']
energy_types = ['Solar', 'Wind', 'Hydro']
# Generate combinations
data = []
for region in regions:
    for energy in energy_types:
        for _ in range(10):  # 10 data points per category
            co2 = random.gauss(mu=5 if energy == 'Hydro' else 3, sigma=1.5)
            data.append([region, energy, round(max(co2, 0.2), 2)])  # Ensure no negative emissions
df = pd.DataFrame(data, columns=['region', 'energy_type', 'co2_emissions'])

"""random.gauss() draws a random number from a normal (bell curve) distribution.
mu -> mean. The centre of the distribution and the EXPECTED co2 value. In the conditional,
it's stating that if the energy type is Hydro, the mean will be 5. The rest will be 3, indicating
Hydro is modeled to have higher average emissions than Solar and Wind.
sigma -> standard deviation. meaning each value can vary roughly ±1.5 units around the mean, simulating real-world variability."""
```

```python
# Step 2: Create new combined category
df['region_energy'] = df['region'] + "_" + df['energy_type']
```

```python
# Step 3: Plot box plot
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='region_energy', y='co2_emissions')
plt.title("CO₂ Emissions by Region + Energy Strategy")
plt.xticks(rotation=45)
plt.ylabel("CO₂ Emissions (tonnes per capita)")
plt.xlabel("Region and Energy Type")
plt.tight_layout()
plt.show()
```
