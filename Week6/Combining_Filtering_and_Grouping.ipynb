```python
import pandas as pd
df = pd.read_csv("week6_grouping_filtering.csv")
print(df.head())
```

```python
# Check for missing values
print(df.isnull().sum())
# Drop rows with missing attendance
df_cleaned = df.dropna(subset=['Attendance'])
print("Rows after cleaning:", len(df_cleaned))
#Question: How many rows were removed? Why does this matter?
"""Means that the data has no missing attendance data. Data is pure."""
```

```python
df_2023 = df_cleaned[df_cleaned['Year'] == 2023]
df_2023.head()
```

```python
#How many rows are left for 2023 only?
print("Rows in 2023 only:", len(df_2023))
```

```python
summary = df_2023.groupby(['Region', 'Gender'])['Attendance'].agg(['mean', 'count']).reset_index()
print(summary)
```

```python
pivot = df_2023.pivot_table(values='Attendance', index='Region', columns='Gender', aggfunc='mean')
print(pivot)
```

```python
region_avg = df_2023.groupby('Region')['Attendance'].mean()
print(region_avg)
```

```python
highest_regions = region_avg[region_avg > 75]
print(f"The highest regions performing best attendance are: {highest_regions}")
```

```python
result = (
    df_cleaned[df_cleaned['Year'] == 2023]
    .groupby(['Region', 'Gender'])['Attendance']
    .agg(['mean', 'count'])
    .reset_index()
)
print(result)
```
