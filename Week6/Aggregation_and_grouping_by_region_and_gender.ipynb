```python
"""You are working for an educational analytics team focused on SDG 4: Quality Education. You’ve been given attendance data across
multiple regions, genders, and years. Your task is to summarize attendance data by region and gender using Python aggregation
techniques."""

import pandas as pd

#Loading dataset
df = pd.read_csv("week6_aggregation_example.csv")
df.head()
```

```python
regional_avg = df.groupby("Region")["Attendance"].mean().reset_index()
print(regional_avg)
```

```python
highest_average_attendance = df.groupby("Region")["Attendance"].agg()
print(f"The region with the highest average attendance is: {highest_average_attendance}")
```
