```python
"""You are working for an educational analytics team focused on SDG 4: Quality Education. You’ve been given attendance data across
multiple regions, genders, and years. Your task is to summarize attendance data by region and gender using Python aggregation
techniques."""

import pandas as pd

#Loading dataset
df = pd.read_csv("week6_aggregation_example.csv")
df.head()
```

```python
regional_avg = df.groupby("Region")["Attendance"].mean().reset_index()
print(regional_avg)
```

```python
#Finding the highest region with average attendance
highest_region = regional_avg.loc[regional_avg["Attendance"].idxmax(), "Region"] #.idxmax returns the index of the row with the higest value
highest_value = regional_avg['Attendance'].max()
print(f"\nRegion with highest average attendance: {highest_region}")
print(f"Average attendance: {highest_value:.2f}")
```

```python
by_region_gender = df.groupby(['Region', 'Gender'])['Attendance'].mean().reset_index()
print(by_region_gender)
```

```python
pivot = df.pivot_table(values='Attendance', index='Region', columns='Gender', aggfunc='mean')
print(pivot)
```

```python
df_2023 = df[df['Year'] == 2023]
avg_2023 = df_2023.groupby(['Region'])['Attendance'].mean().reset_index()
print(avg_2023)
```

```python
# Example: High attendance female students
filtered = df[(df['Gender'] == 'Female') & (df['Attendance'] > 85)]
summary = filtered.groupby('Region')['Attendance'].mean()
print(summary)
```
