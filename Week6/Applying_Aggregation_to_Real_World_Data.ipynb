```python
import pandas as pd
df = pd.read_csv("week6_grouping_filtering.csv")
df.head()
```

```python
df_cleaned = df.dropna(subset=['Attendance'])
```

```python
region_year = (
    df_cleaned.groupby(['Region', 'Year'])['Attendance']
    .mean()
    .reset_index()
)
print(region_year) #helps assess progress or regression in school attendance year-to-year, and is useful in time-based reporting.
```

```python
gender_region = (
    df_cleaned.groupby(['Region', 'Gender'])['Attendance']
    .mean()
    .reset_index()
)
print(gender_region)
```

```python
#Look for large differences between male and female attendance — a potential sign of inequality that could be targeted for
#improvement.
```

```python
region_avg = df_cleaned.groupby('Region')['Attendance'].mean()
print("High Attendance:\n", region_avg[region_avg > 75])
print("Low Attendance:\n", region_avg[region_avg < 75])
#This kind of filter helps surface which regions may need further support, which aligns with monitoring goals under SDG 4
```
