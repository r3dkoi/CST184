```r
#You surveyed 50 households after implementing a clean water project. 38 households report access to safe water. You want to estimate the probability that more than 70% of the community has access overall.
```

```r
#Step 1 – Define a Prior (Beta Distribution)

#Use a Beta(2, 2) prior — this assumes no strong prior belief and represents a flat, neutral view.

# Neutral prior: Beta(2, 2)
alpha_prior <- 2
beta_prior <- 2
```

```r
#Step 2 – Add the Observed Data

# Observed data
successes <- 38
failures <- 12 #50 - 38 = 12
```

```r
#Step 3 – Update to Get Posterior

#In a beta-binomial model, the posterior is:

#Posterior = Beta(α + successes, β + failures)

# Posterior parameters
alpha_post <- alpha_prior + successes
beta_post <- beta_prior + failures
```

```r
#Step 4 – Simulate Posterior Samples

#Simulate 10,000 posterior values from the updated beta distribution:

posterior <- rbeta(10000, alpha_post, beta_post)
posterior
```

```r
#Step 5 – Visualize the Posterior Distribution

hist(posterior, breaks = 50, col = "skyblue",
     main = "Posterior Distribution",
     xlab = "Proportion of Households with Access")
```

```r
#Step 6 – Estimate Success Probability

#What is the probability that access to water exceeds 70%?

mean(posterior > 0.7)
#76% chance that the program reached at least 70% of the community.
```
