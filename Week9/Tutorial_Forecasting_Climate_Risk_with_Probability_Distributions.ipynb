```r
#Scenario
print("You are a junior data analyst working with a climate research team supporting SDG 13: Climate Action. Your task is to model and visualize the likelihood of climate-related events to inform policy decisions for the coming year. You’ll use data related to extreme weather events and temperature anomalies to:")
print("Model the number of events using discrete distributions (binomial and Poisson)")
print("Estimate expected values for risk forecasting")
print("Model temperature anomalies with a normal distribution")
print("Visualize all of your results using ggplot2")
```

```r
#Step 1 - Setup
install.packages("ggplot2")
library(ggplot2)
```

```r
#Step 2 – Binomial Distribution: Community Preparedness
#A climate  response team surveys 20 communities. 
#Each has a 65% chance of being adequately prepared for floods. 
#Use a binomial distribution to model the number of prepared communities.

#Defining parameters
n <- 20
p <- 0.65
x_binom <- 0:n
y_binom <- dbinom(x_binom, size = n, prob = p)
expected_binom <- n * p

#Creating dataframe
df_binom <- data.frame(Prepared = x_binom, Probability = y_binom)

#plotting the graph
ggplot(df_binom, aes(Prepared, Probability)) +
  geom_bar(stat = "identity", fill = "steelblue") +
  geom_vline(xintercept = expected_binom, linetype = "dashed", color = "red") +
  labs(
    title = "Binomial Distribution: Community Flood Preparedness",
    subtitle = "Expected Value (dashed red line)",
    x = "Number of Prepared Communities",
    y = "Probability"
  ) +
  theme_minimal()
```

```r
#Step 2.5 Modified distribution shift
n <- 20
p <- 0.4
x_binom <- 0:n
y_binom <- dbinom(x_binom, size = n, prob = p)
expected_binom <- n * p

df_binom <- data.frame(Prepared = x_binom, Probability = y_binom)

ggplot(df_binom, aes(Prepared, Probability)) +
  geom_bar(stat = "identity", fill = "steelblue") +
  geom_vline(xintercept = expected_binom, linetype = "dashed", color = "red") +
  labs(
    title = "Modified Binomial Distribution: Community Flood Preparedness",
    subtitle = "Expected Value (dashed red line)",
    x = "Number of Prepared Communities",
    y = "Probability"
  ) +
  theme_minimal()
```

```r
#Step 3 -  Poisson Distribution: Forecasting Storm Events
#A coastal region averages 5 extreme storms per year. 
#Use a Poisson distribution to model how many storms might occur next year.

#Defining parameters
lambda <- 5
x_poisson <- 0:15
y_poisson <- dpois(x_poisson, lambda = lambda)

#Creating dataframe
df_poisson <- data.frame(Events = x_poisson, Probability = y_poisson)

#Visualising graph
ggplot(df_poisson, aes(Events, Probability)) +
  geom_bar(stat = "identity", fill = "firebrick") +
  geom_vline(xintercept = lambda, linetype = "dashed", color = "black") +
  labs(
    title = "Poisson Distribution: Storm Frequency Forecast",
    subtitle = "Expected Number of Events = λ",
    x = "Number of Storms",
    y = "Probability"
  ) +
  theme_minimal()
```

```r
#Step 4 - Normal Distribution: Temperature Anomalies
#Average yearly temperature anomaly in your region is 1.1°C, with a standard deviation of 0.4°C. 
#Visualize how temperature anomalies are distributed across years.

#Defining parameters
mean_temp <- 1.1
sd_temp <- 0.4
x_temp <- seq(-0.5, 2.5, length.out = 200)
y_temp <- dnorm(x_temp, mean = mean_temp, sd = sd_temp)

#Creating dataframe
df_temp <- data.frame(Anomaly = x_temp, Density = y_temp)

#Visualising the graph
ggplot(df_temp, aes(Anomaly, Density)) +
  geom_line(color = "darkgreen", linewidth = 1.2) +
  geom_vline(xintercept = mean_temp, linetype = "dashed", color = "blue") +
  labs(
    title = "Normal Distribution: Temperature Anomalies",
    subtitle = "Mean = 1.1°C (dashed line)",
    x = "Temperature Anomaly (°C)",
    y = "Density"
  ) +
  theme_minimal()
```
