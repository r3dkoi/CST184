```r
#Goal: Use R to generate and visualize normal distributions
#using dnorm() and ggplot2. 
#Modify the mean and standard deviation to see how the curve changes. 
#Apply this understanding to real-world continuous data scenarios like SDG-aligned climate data and income levels.
```

```r
#Step 1 -- Setting up the R notebook
install.packages("ggplot2")
library(ggplot2)
```

```r
#Step 2 - Generating a normal distribution
#Simulating  normal distribution representing daily temperature anomalies (°C) in a region aligned with SDG 13.

# Define parameters
mean_temp <- 0     # average anomaly = 0°C. Curve is centred at 0.
sd_temp <- 1       # standard deviation = 1°C. The spread i 1 on either side.

# Create x-values for the curve
x <- seq(-4, 4, length.out = 200) #seq() generates a sequence of 200 evenly spaced numbers between -4 and +4. They are the points along the X axis.

# Generate y-values using the density function
y <- dnorm(x, mean = mean_temp, sd = sd_temp) #For each x value, it calculates how "likely" or "dense" the point is. Higher y = more probable temp. anomaly

# Create data frame
df <- data.frame(x, y)

# Plot the normal curve
ggplot(df, aes(x, y)) + #Tells ggplot which data and axes to use
  geom_line(color = "steelblue", linewidth = 1.2) + #Draws the continuous bell curve line
  labs(
    title = "Normal Distribution of Temperature Anomalies",
    x = "Temperature Anomaly (°C)",
    y = "Density"
  ) + #Adds title and axis labels
  theme_minimal() #The style of the actual graph
```

```r
#Step 3 - Modify the Mean and Spread 
#Line graph
# Example: higher mean and wider spread
mean_temp <- 1.5
sd_temp <- 2

x <- seq(-6, 6, length.out = 200)
y <- dnorm(x, mean = mean_temp, sd = sd_temp)
df <- data.frame(x, y)

ggplot(df, aes(x, y)) +
  geom_line(color = "firebrick", size = 1.2) +
  labs(
    title = "Modified Normal Distribution (Mean = 1.5, SD = 2)",
    x = "Temperature Anomaly (°C)",
    y = "Density"
  ) +
  theme_minimal()
```

```r
#Step 4 - Interpret the Curve
print("Where is the peak of the curve located?")
print("The peak of the curve is located at 0.20 density and 1.5 degrees celsius.")
```

```r
print("How wide is the range that most values fall into?")
#Formula: Lower bound = Mean - (? × SD) to Upper bound = Mean + (? × SD)
lower_bound <- mean_temp - (2*sd_temp)
upper_bound <- mean_temp + (2*sd_temp)
range <- upper_bound - lower_bound
print(range)
```

```r
print("What does this tell you about variability in your data?")
print("A larger SD means temperature anomalies are more spread out and variable — temperatures are less predictable and deviate more from the average.")
```
