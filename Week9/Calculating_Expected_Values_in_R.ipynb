```r
#Step 1 - Setup
install.packages("ggplot2")
library(ggplot2)
```

```r
#Step 2 - Binomial distribution: Health survey
#Scenario (SDG 3): In a survey of 10 people, 
#each has a 70% chance of having received healthcare support. 
#What is the expected number of people to say “yes”?""

#Defining parameters
n <- 10 #total participants
p <- 0.7 #probability of success

#Calculating expected value
expected_binom <- n * p
paste("The expected no. of people to say 'yes' in the survey is", expected_binom)
```

```r
#Step 2.5 Chnging the probability of success
#Defining parameters
n <- 10 #total participants
p <- 0.9 #probability of success

#Calculating expected value
expected_binom <- n * p
paste("The expected no. of people to say 'yes' in the survey is", expected_binom)
```

```r
#Step 3 - Poisson Distribution: Extreme Weather
#Scenario (SDG 13): A region averages 4 extreme weather events per year. 
#What is the expected number of events next year?

#Define Poisson rate (lambda)
lambda <- 4
#Expected Poisson value
expected_poisson <- lambda
paste("The expected extreme weather events next is", expected_poisson)
```

```r
#Optional - Visualising the Binomial Distribution
binom_df <- data.frame(
  successes = 0:n, #generates every whole number from 0 to n (e.g. 0,1,2...10)
  probability = dbinom(0:n, size = n, prob = p) #calculates the exact probability of each number of successes
)

# Plot
ggplot(binom_df, aes(x = successes, y = probability)) +
  geom_bar(stat = "identity", fill = "steelblue") +  #tells R to use the y values as-is rather than counting them
  geom_vline(xintercept = expected_binom, linetype = "dashed", color = "red") + #geom_vline draws a vertical line. #xintercept = expected_binom places it at the expected value 
  labs(
    title = "Binomial Distribution with Expected Value",
    x = "Number of People Saying Yes",
    y = "Probability"
  ) +
  theme_minimal()

#This will display the distribution and highlight the expected value with a red dashed line.
```
