```python
#Scenario (SDG 3 – Health):

"""In a survey of 10 people, each has a 60% chance of having received healthcare support. What is the probability that exactly 7 people said yes?

We can model this using a binomial distribution: fixed number of trials (10), fixed probability of success (0.6)."""
```

```python
#P(X = k) = (n choose k) x pk x (1-p)n-k
#n choose k formula = n! / (k! x (n-k)1)
```

```python
n = 10 #Number of fixed trials
p = 0.6 #Fixed probability of 6
k = 7 #(Want 7 people to say yes)
```

```python
import math

#Step 1 Calculate n choose k

n_choose_k = math.factorial(10) / (math.factorial(7) * math.factorial(3))
print(n_choose_k)
```

```python
p_say_yes_1 = n_choose_k * (0.6)**7 * (0.4)**3 * 100
print(round(p_say_yes_1, 2))
print(f"The probability of exactly 7 people saying yes is: {round(p_say_yes_1, 2)}%")
```

```python
#Scenario (SDG 13 – Climate):
#A coastal region experiences an average of 4 extreme weather events per year. What’s the probability that exactly 6 events happen this year?
#This is a typical Poisson situation — we know the average rate (λ = 4), and we want the probability of a specific count (6).
```

```python
# P(X = k) = (λ^k × e^-λ) / k!
```

```python
λ = 4
k = 6
e = 2.71828
```

```python
p_extreme_weather_per_year = ((λ**6 * e**-λ) / math.factorial(6)) * 100
print(f"The probability of exactly 6 extreme weather events happening this year: {round(p_extreme_weather_per_year, 2)}%")
```
