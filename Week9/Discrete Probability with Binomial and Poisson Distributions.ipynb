```r
#Step 1 Setup
install.packages("ggplot2")
library(ggplot2)
```

```r
#Step 2 - Binomial Distribution Example
#Scenario: In an SDG 3 health survey, each person has a 60% chance of reporting they received care. 
#What is the probability of getting exactly 0–10 “yes” responses in a group of 10?

# Create binomial distribution data
n <- 10     # number of trials
p <- 0.6  # probability of success

binom_df <- data.frame(
  successes = 0:n,
  probability = dbinom(0:n, size = n, prob = p)
)

# Plot the distribution
ggplot(binom_df, aes(x = successes, y = probability)) +
  geom_bar(stat = "identity", fill = "steelblue") +
  labs(
    title = "Binomial Distribution (n = 10, p = 0.6)",
    x = "Number of Successes",
    y = "Probability"
  ) +
  theme_minimal()

```

```r
#Step 3 Poisson Distribution Example
#Scenario: In a coastal SDG 13 region, the average number of extreme storms per year is 4. 
#What does the distribution of yearly storm counts look like?
# Create Poisson distribution data
lambda <- 4

poisson_df <- data.frame(
  events = 0:15,
  probability = dpois(0:15, lambda = lambda)
)

# Plot the distribution
ggplot(poisson_df, aes(x = events, y = probability)) +
  geom_bar(stat = "identity", fill = "firebrick") +
  labs(
    title = "Poisson Distribution (λ = 4)",
    x = "Number of Events",
    y = "Probability"
  ) +
  theme_minimal()
```
