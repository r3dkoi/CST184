```r
#Step 1 Set-up
install.packages("ggplot2")
library(ggplot2)
```

```r
#Step 2 Binomial Distribution Example
#Scenario: In a group of 10 people, each has a 70% chance of having received a vaccine.
#What you’ll do: Simulate the probability of each possible number of “yes” responses.

#Defining parameters
n <- 10
p <- 0.7
x <- 0:n
y <- dbinom(x, size = n, prob = p)

#Creating a dataframe
binom_df <- data.frame(
    successes = x,
    probability = y 
    )
expected_binom = n * p
```

```r
#Plotting the bar graph
ggplot(binom_df, aes(x = successes, y = probability)) +
  geom_bar(stat = "identity", fill = "steelblue") +
  geom_vline(xintercept = expected_binom, linetype = "dashed", color = "red") +
  labs(
    title = "Binomial Distribution with Expected Value",
    x = "Number of People Saying Yes",
    y = "Probability"
  ) +
  theme_minimal()
```

```r
#Step 3 Poisson Distribution Example
#A coastal region experiences an average of 4 extreme weather events per year. What’s the distribution of outcomes?

#Defining the parameters
lambda <- 4
x <- 0:15
y <- dpois(x, lambda = lambda)

#Creating a dataframe
poisson_df <- data.frame(
    successes = x,
    probability = y
    )
expected_poisson = lambda

#Plotting graph
    ggplot(poisson_df, aes(x = successes, y = probability)) +
  geom_bar(stat = "identity", fill = "firebrick") +
  geom_vline(xintercept = expected_poisson, linetype = "dashed", color = "black") +
  labs(
    title = "Poisson Distribution with Expected Value",
    x = "Number of extreme weather events per year",
    y = "Probability"
  ) +
  theme_minimal()
```

```r
#Step 4 Normal Distribution visualisation
#A country’s average daily calorie intake is 2300 kcal with a standard deviation of 400.
#What does this look like across the population?

#Defining parameters
mean_calories <- 2300
calories_sd <- 400
expected_average <- mean_calories

#Creating dataframe
x <- seq(1000, 3500, length.out = 200)
y <- dnorm(x, mean = mean_calories, sd = calories_sd)
normal_df <- data.frame(x, y)

ggplot(normal_df, aes(x, y)) +
  geom_line(color = "green", linewidth = 1.2) +
  labs(
    title = "A country's normal distribution calorie intake (Mean = 2300, SD = 400)",
    x = "Daily calorie intake (kcal)",
    y = "Density"
  ) +
  theme_minimal()

#The population is **represented by the whole curve**
```
