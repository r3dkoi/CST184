```r
"""You are working with a global research group analyzing climate change indicators for SDG 13: Climate Action. The team has collected three metrics for a sample of countries:

    CO₂ emissions per capita (metric tons)
    Mean temperature anomaly (°C above baseline)
    Deforestation rate (% of forest area lost)

Your task is to use matrix transformations in R to standardize this data and prepare it for comparative analysis across countries."""
```

```r
# Create a 5x3 matrix: [CO2, Temp Anomaly, Deforestation Rate]
climate <- matrix(c(
  9.2, 1.1, 0.8,   # Country A
  6.5, 0.9, 1.2,   # Country B
  2.3, 0.7, 2.5,   # Country C
  3.8, 1.0, 1.5,   # Country D
  7.1, 1.2, 0.9    # Country E
), nrow=5, byrow=TRUE)

colnames(climate) <- c("CO2", "Temp", "Deforestation")
rownames(climate) <- c("A", "B", "C", "D", "E")

print(climate)
```

```r
scaled_climate <- scale(climate)
print(scaled_climate)

#Try adding a new country with extreme values and observe how the scaling changes.
```

```r
# Transformation matrix: keep CO2 and Temp, invert Deforestation
flip <- matrix(c(1, 0, 0,
                 0, 1, 0,
                 0, 0, -1), nrow=3)

transformed <- scaled_climate %*% flip
print(transformed)

#Question: Why might inverting some metrics make sense when analyzing sustainability data?
```

```r
transposed <- t(transformed)
print(transposed)
```

```r
#Adding a sixth country to the matrix
# Create a 5x3 matrix: [CO2, Temp Anomaly, Deforestation Rate]
climate <- matrix(c(
  9.2, 1.1, 0.8,   # Country A
  6.5, 0.9, 1.2,   # Country B
  2.3, 0.7, 2.5,   # Country C
  3.8, 1.0, 1.5,   # Country D
  7.1, 1.2, 0.9,    # Country E
10.0, 9.9, 5.5 #Country F
), nrow=6, byrow=TRUE)

colnames(climate) <- c("CO2", "Temp", "Deforestation")
rownames(climate) <- c("A", "B", "C", "D", "E", "F")

print(climate)
```

```r
scaled_climate <- scale(climate)
print(scaled_climate)
```

```r
# Change the transformation matrix to weight CO₂ more heavily (e.g., multiply it by 2).
flip <- matrix(c(1, 0, 0,
                 0, 1, 0,
                 0, 0, -1), nrow=3)

flip_weighted <- matrix(c(-2, 0, 0,
                          0, 1, 0,
                          0, 0 , -1), nrow=3) #-2 inverts + adds double weight to CO2 column

transformed <- scaled_climate %*% flip_weighted
print(transformed)

#Question: Why might inverting some metrics make sense when analyzing sustainability data?
When doing PCA or similarity analysis, higher = better is assumed by default. But some metrics are the opposite — higher is worse. Inverting realigns them so the math reflects reality
```

```r
transposed <- t(transformed)
print(transposed)
```

```r
scaled <- scale(climate)
cor_matrix <- cor(scaled)
print(cor_matrix)

#Prompt: Which variables are most positively correlated? Which are most negatively correlated? What might this tell us about climate patterns?
```
