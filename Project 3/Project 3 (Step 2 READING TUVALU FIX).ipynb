```r
#Purpose
#analysing rising sea level trends in the Pacific and assessing the projected impact on low-lying nations

#Step 0 Importing Libraries
library(tidyverse)
library(ggplot2)
library(reshape2)
```

```r
#Step 1 Importing dataset and inspection
main_df <- read.csv("Tuvalu_sea_level_trend.csv")
```

```r
#There was metadata in the original CSV file, so I just removed the metadata text in order to make the CSV be read properly.
main_df <- read.csv("Tuvalu_sea_level_trend.csv", check.names = FALSE, row.names = NULL, fill = TRUE) # check.names = FALSE prevents R from modifying column names automatically
# row.names = NULL forces R to use default integer row numbering
# fill = TRUE handles rows with unequal field counts caused by trailing commas

main_df <- main_df[, 1:6]  # The trailing comma on every data row created an empty 7th column of NAs.
# Subsetting to columns 1:6 removes it, keeping only the six meaningful columns.
```

```r
glimpse(main_df) #shows every column name and it's data type and preview of actual values. Used instead of head() and str().
```

```r
colnames(main_df) <- c("Year", "Month", "Monthly_MSL", "Linear_Trend", "High_Conf", "Low_Conf")  # Manually reassigning all six column names corrects the shift caused by
# row.names = NULL introducing an extra leading column during import.
glimpse(main_df)
```

```r
main_df$Year <- as.integer(main_df$Year)
# Year was read as <chr> due to the import issues above.
# Converting to integer ensures it is treated as a numeric variable,
# which is required for time series plotting and calculations later.
```

```r
glimpse(main_df)
# Final inspection confirms all six columns are correctly named, Year is now
# <int>, and all other columns hold the expected numeric (dbl) values.
```

```r
write.csv(main_df, "fixed_cleaned_Tuvalu_sea_level_clean.csv", row.names = FALSE)
```
