```python
import pandas as pd

df = pd.read_csv("renewable_energy.csv")
df.head()
```

```python
#Printing only Country and %Renewable Colums
df["%Renewable"]
```

```python
df["Country"]
```

```python
#Filtering the Data
high_renewable = df[df["%Renewable"] > 50]
high_renewable
```

```python
#Filtering the Data
high_renewable = df[df["%Renewable"] > 70]
high_renewable
```

```python
#Creating a Set
high_set = set(high_renewable["Country"]) #Creates a set that has the boolean mask of only showing the rows that have a >50 value in them. [Country] will only print print the countries with >50 renewable energy.

print("Countries with the >50:", high_set)
print("Number of countries stated above are:", len(high_set))
```

```python
#Creating a Dictionary
emissions_dict = dict(zip(df["Country"], df["CO2_Emissions"])) #Zip function takes corresponding elements from BOTH series of Country and CO2_Emissions and pairs them together 
print(emissions_dict["Brazil"])
```

```python
#Bar Chart of CO2 Emissions
import matplotlib.pyplot as plt

plt.bar(emissions_dict.keys(), emissions_dict.values())
plt.xticks(rotation=45)
plt.title("CO₂ Emissions per Capita by Country")
plt.ylabel("Tonnes")
plt.grid(True)
plt.bar(df["Country"], df["CO2_Emissions"], color='green') 
plt.show()
```

```python
#Scatterplot - Population vs Renewable
plt.scatter(df["Population"], df["%Renewable"])

for i in range(len(df)):
    plt.text(df["Population"][i], df["%Renewable"][i], df["Country"][i])

plt.title("Population vs % Renewable Energy Use")
plt.xlabel("Population (millions)")
plt.ylabel("% Renewable")
plt.grid(True)
plt.show()
```
