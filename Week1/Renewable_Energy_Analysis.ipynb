```python
import pandas as pd

df = pd.read_csv("renewable_energy.csv")
df.head()
```

```python
#Printing only Country and %Renewable Colums
df["%Renewable"]
```

```python
df["Country"]
```

```python
#Filtering the Data
high_renewable = df[df["%Renewable"] > 50]
high_renewable
```

```python
#Filtering the Data
high_renewable = df[df["%Renewable"] > 70]
high_renewable
```

```python
#Creating a Set
high_set = set(high_renewable["Country"]) #Creates a set that has the boolean mask of only showing the rows that have a >50 value in them. [Country] will only print print the countries with >50 renewable energy.

print("Countries with the >50:", high_set)
print("Number of countries stated above are:", len(high_set))
```

```python
#Creating a Dictionary
emissions_dict = dict(zip(df["Country"], df["CO2_Emissions"])) #Zip function takes corresponding elements from BOTH series of Country and CO2_Emissions and pairs them together 
print(emissions_dict["Brazil"])
```

```python
#Bar Chart of CO2 Emissions
import matplotlib.pyplot as plt

plt.bar(emissions_dict.keys(), emissions_dict.values(), color='green')
plt.xticks(rotation=45)
plt.title("CO₂ Emissions per Capita by Country")
plt.ylabel("Tonnes")
plt.grid(True)
plt.show()
```

```python
#Scatterplot - Population vs Renewable
plt.scatter(df["Population"], df["%Renewable"])

for i in range(len(df)):
    plt.text(df["Population"][i], df["%Renewable"][i], df["Country"][i])

plt.title("Population vs % Renewable Energy Use")
plt.xlabel("Population (millions)")
plt.ylabel("% Renewable")
plt.grid(True)
plt.show()
```

```python
#Plotting only countries in Asia
# asia_countries_df = pd.DataFrame({
    #'Vietnam': [64],
    #'India': [24])} #Scratch this I think I'm gonna do something different

# asia_countries_df = dict(zip(df["Country"], df["Region"], df["%Renewable"])) #But then how do I just seperate it to Asia country?


#Filter for Asia Countries first
asia_df = df[df["Region"] == "Asia"] #Filters only rows that have Asia under the Region label/column
plt.scatter(asia_df["Population"], asia_df["%Renewable"]) 
for i in range(len(asia_df)):
    plt.text(asia_df["Population"].iloc[i], asia_df["%Renewable"].iloc[i], asia_df["Country"].iloc[i]) #Accesses position of INDEX. Consistent :D

plt.title("Population vs % Renewable Energy Use (Asia)")
plt.xlabel("Population (millions)")
plt.ylabel("% Renewable")
plt.grid(True)
plt.show()
```

```python
subset = df[df["%Renewable"] > 60]
subset.to_csv("high_renewable_countries.csv", index=False)
```

```python
import matplotlib.pyplot as plt

#Which regions have the highest renewable energy adoption?
#Breaking it down, I can make a scatter graph of using X as the Regions and Y as the Renewable energy 
plt.scatter(df["Region"], df["%Renewable"])

for i in range(len(df)):
    plt.text(df["Region"][i], df["%Renewable"][i], df["Country"][i])

plt.title("Population vs % Renewable Energy Use")
plt.xlabel("Population (millions)")
plt.ylabel("% Renewable")
plt.grid(True)
plt.show()
```

```python
highest_renewable_dict = dict(zip(df["Region"], df["%Renewable"]))

plt.bar(highest_renewable_dict.keys(), highest_renewable_dict.values(), color="green")
plt.xticks(rotation=45)
plt.title("Highest Regions with renewable energy adoption")
plt.ylabel("%")
plt.grid(True)
plt.show()
```
