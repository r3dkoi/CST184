```r
#In a single pipeline for each condition, find all flights that meet the condition:
glimpse(flights)
```

```r
flights |> #Flights firrst 
    filter(dep_delay >= 120) |> #then flights that have been delayed for more than 2 or more hours
    filter(dest %in% c("IAH", "HOU")) |> #then have flew to IAH or HOU
    filter(carrier %in% c("UA", "AA","DL")) |> #then operated by United, American, or Delta
    filter(month %in% c(7, 8, 9)) |> #%in% is used because you want to combine == AND | / #then departed in sumer
    filter(arr_delay >= 120 & dep_delay <= 0) |> #then arrived more than twoo hours late but didn't leave late
    filter(dep_delay == 60 & dep_delay - arr_delay > 30)
    
    
```

```r
flights |> #Flights first 
    filter(dep_delay >= 120) 
```

```r
flights |> arrange(desc(dep_delay))  #Longest departure delays
```

```r
flights |> arrange(dep_delay) #Earliest departures in the morning
```
